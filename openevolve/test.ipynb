```python
import xml.etree.ElementTree as ET
import random

# 您提供的 XML 字符串
xml = '''
        <mujoco model="ant">
        <compiler angle="degree" coordinate="global" inertiafromgeom="true"/>
        <option integrator="RK4" timestep="0.01"/>
        <default>
            <joint armature="1" damping="1" limited="true"/>
            <geom conaffinity="0" condim="3" density="5.0" friction="1 0.5 0.5" margin="0.01" rgba=".98 .54 .56 1."/>
        </default>
        <visual>
            <headlight ambient=".1 .1 .1" diffuse=".6 .6 .6" specular="0.3 0.3 0.3"/>
            <map znear=".01"/>
            <quality shadowsize="16384"/>
        </visual>
        <worldbody>
            <light cutoff="100" diffuse="1 1 1" dir="-0 0 -1.3" directional="true" exponent="1" pos="0 0 1.3" specular=".1 .1 .1"/>
            <geom conaffinity="1" condim="3" name="floor" pos="0 0 0" rgba="0.8 0.9 0.8 1" size="200 200 .125" type="plane" material="grid_new"/>
            <body name="0" pos="0 0 0">
            <camera name="track" mode="trackcom" pos="0 -3 0.3" xyaxes="1 0 0 0 0 1"/>
            <geom pos="0 0 0" size="0.25" type="sphere"/>
            <joint armature="0" damping="0" limited="false" margin="0.01" name="root" pos="0 0 0" type="free"/>
            <body name="1" pos="0 0 0">
                <joint axis="-0.707107 0.707107 0" name="1_joint" pos="0 0 0" range="-30 30" type="hinge"/>
                <geom fromto="0 0 0 0.4 0.4 0" size="0.08" type="capsule"/>
            </body>
            <body name="2" pos="0 0 0">
                <joint axis="-0.707107 -0.707107 0" name="2_joint" pos="0 0 0" range="-30 30" type="hinge"/>
                <geom fromto="0 0 0 -0.4 0.4 0" size="0.08" type="capsule"/>
            </body>
            <body name="3" pos="0 0 0">
                <joint axis="0.707107 -0.707107 0" name="3_joint" pos="0 0 0" range="-30 30" type="hinge"/>
                <geom fromto="0 0 0 -0.4 -0.4 0" size="0.08" type="capsule"/>
            </body>
            <body name="4" pos="0 0 0">
                <joint axis="0.707107 0.707107 0" name="4_joint" pos="0 0 0" range="-30 30" type="hinge"/>
                <geom fromto="0 0 0 0.4 -0.4 0" size="0.08" type="capsule"/>
            </body>
            </body>
        </worldbody>
        <actuator>
            <motor ctrllimited="true" ctrlrange="-1.0 1.0" joint="1_joint" gear="150" name="1_joint"/>
            <motor ctrllimited="true" ctrlrange="-1.0 1.0" joint="2_joint" gear="150" name="2_joint"/>
            <motor ctrllimited="true" ctrlrange="-1.0 1.0" joint="3_joint" gear="150" name="3_joint"/>
            <motor ctrllimited="true" ctrlrange="-1.0 1.0" joint="4_joint" gear="150" name="4_joint"/>
        </actuator>
        <asset>
            <texture type="skybox" builtin="gradient" rgb1=".4 .5 .6" rgb2="0 0 0" width="100" height="100"/>
            <texture builtin="flat" height="1278" mark="cross" markrgb="1 1 1" name="texgeom" random="0.01" rgb1="0.8 0.6 0.4" rgb2="0.8 0.6 0.4" type="cube" width="127"/>
            <texture builtin="checker" height="100" name="texplane" rgb1="0 0 0" rgb2="0.8 0.8 0.8" type="2d" width="100"/>
            <material name="MatPlane" reflectance="0.5" shininess="1" specular="1" texrepeat="60 60" texture="texplane"/>
            <material name="geom" texture="texgeom" texuniform="true"/>
            <texture name="grid_new" type="2d" builtin="checker" rgb1=".1 .3 .2" rgb2=".2 .4 .3" width="1000" height="1000" mark="none" markrgb=".8 .6 .4"/>
            <material name="grid_new" texture="grid_new" texrepeat="0.2 0.2" texuniform="true" reflectance=".2"/>
        </asset>
        </mujoco>
    '''

def generate_and_print_control_strategy(xml_string, num_frames=10):
    """
    解析 MuJoCo XML，识别可动关节，并生成并打印一个随机的控制策略。
    """
    # 1. 解析 XML 字符串
    root = ET.fromstring(xml_string)

    # 2. 查找所有类型不为 "free" 的可动关节
    #    "free" 关节代表模型整体的自由度，不是由电机驱动的
    movable_joints = [
        joint.get('name') for joint in root.findall('.//joint')
        if joint.get('type') != 'free' and joint.get('name')
    ]
    num_movable_joints = len(movable_joints)

    # 3. 生成随机控制策略
    control_strategy = []
    for _ in range(num_frames):
        # 为每个关节生成一个 [-1.0, 1.0] 之间的随机浮点数
        frame_controls = [random.uniform(-1.0, 1.0) for _ in range(num_movable_joints)]
        control_strategy.append(frame_controls)

    # 4. 将策略转化为可视的文本形式并打印
    print("==========================================================")
    print("          MuJoCo 机器人随机运动控制策略")
    print("==========================================================")
    print(f"✅ 已识别的可动关节数量: {num_movable_joints}")
    print(f"✅ 关节名称: {movable_joints}")
    print("----------------------------------------------------------")

    # 打印表头
    header = " Frame |" + " | ".join([f"{name:^11}" for name in movable_joints])
    print(header)
    print("----------------------------------------------------------")

    # 逐帧打印控制数据
    for i, frame in enumerate(control_strategy):
        # 格式化输出：+号表示显示正负号，10表示总宽度，.4f表示4位小数
        control_values = " | ".join([f"{val: >+10.4f}" for val in frame])
        # 格式化输出：<5 表示左对齐，宽度为5
        print(f"  {i+1:<5} | {control_values} ")

    print("==========================================================")


# --- 执行代码 ---
generate_and_print_control_strategy(xml)

# plot the robot from the xml
import mujoco
import mujoco.viewer

model = mujoco.MjModel.fromstring(xml)
data = mujoco.MjData(model)

mujoco.mj_step(model, data)

mujoco.viewer.launch_passive(model, data)
```
